```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
filepath="/content/adult_dataset.csv"
df=pd.read_csv(filepath)
df.head()
df
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
columns = ['age', 'workclass', 'occupation','hours-per-week','native-country']

for col in columns:
    sns.histplot(df[col], kde=True)
    plt.title(f'Histogram of {col}')
    plt.show()
```

```python
df['hours-per-week'] = df['hours-per-week'].fillna(df['hours-per-week'].median())
```

```python
df['age'] = df['age'].fillna(df['age'].median())
```

```python
df['workclass'] = df['workclass'].fillna(df['workclass'].mode()[0])
```

```python
df['occupation'] = df['occupation'].fillna(df['occupation'].mode()[0])
```

```python
df['native-country'] = df['native-country'].fillna(df['native-country'].mode()[0])
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.replace('?', np.nan, inplace=True)
```

encoding

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
```

```python
num_cols = df.select_dtypes(include=["int64", "float64"]).columns

cat_cols = df.select_dtypes(include=["object"]).columns

print("Numerical:", list(num_cols))
print("Categorical:", list(cat_cols))
```

```python
num_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

cat_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("num", num_pipeline, num_cols),
    ("cat", cat_pipeline, cat_cols)
])
```

```python
X = df

X_scaled = preprocessor.fit_transform(X)
```

```python
Q1 = df['age'].quantile(0.25)
Q3 = df['age'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df['age'] = df['age'].clip(lower, upper)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

numerical_cols = ['age', 'fnlwgt', 'education-num', 'capital-gain',
                  'capital-loss', 'hours-per-week']

df[numerical_cols] = scaler.fit_transform(df[numerical_cols])
```

dimensionality reduction

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)
```

```python
import matplotlib.pyplot as plt

plt.scatter(X_pca[:, 0], X_pca[:, 1])
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('PCA Visualization')
plt.show()
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

```python
inertia = []

for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)

plt.plot(range(2, 11), inertia, marker='o')
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method")
plt.show()
```

```python
silhouette_scores = []

# Use only 2000 rows
X_sample = X_scaled[:2000]

for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(X_sample)

    score = silhouette_score(X_sample, labels)
    silhouette_scores.append(score)

print(silhouette_scores)
```

```python
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)

df['Cluster'] = kmeans.fit_predict(X_scaled)

df.head()
```

```python
plt.figure(figsize=(8, 6))

plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    c=df['Cluster'],
    cmap='viridis',
    s=30
)

plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.title("K-Means Clusters")
plt.colorbar(label="Cluster")

plt.show()
```

```python
print(pca.explained_variance_ratio_)
print(pca.explained_variance_ratio_.sum())
```

```python
from sklearn.cluster import AgglomerativeClustering
```

```python
X_sample = X_scaled[:2000].toarray()

agg = AgglomerativeClustering(n_clusters=4)

agg_labels = agg.fit_predict(X_sample)
```

```python
score = silhouette_score(X_sample, agg_labels)

print("Silhouette Score:", score)
```

```python
pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_sample)

plt.figure(figsize=(8, 6))

plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    c=agg_labels,
    cmap='viridis',
    s=30
)

plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.title("Agglomerative Clustering")
plt.colorbar(label="Cluster")
plt.show()
```

K-Means produced a silhouette score of approximately 0.13, while Agglomerative Clustering produced approximately 0.10. Both methods show relatively low cluster separation.

interpretation and profiling

```python
for cluster in sorted(df['Cluster'].unique()):
    print("\n========== Cluster", cluster, "==========")

    cluster_data = df[df['Cluster'] == cluster]

    for col in columns:
        print("\n", col)
        print(
            cluster_data[col]
            .value_counts(normalize=True)
            .head(3)
            .mul(100)
            .round(1)
        )
```

```python
for cluster in sorted(df['Cluster'].unique()):
    cluster_data = df[df['Cluster'] == cluster]

    print(
        "Cluster", cluster,
        "Average age:", round(cluster_data['age'].mean(), 2)
    )
```

```python
pd.crosstab(
    df['Cluster'],
    df['income'],
    normalize='index'
).round(3) * 100
```

```python
income_cluster = pd.crosstab(
    df['Cluster'],
    df['income'],
    normalize='index'
) * 100

income_cluster.plot(
    kind='bar',
    figsize=(8, 5)
)

plt.xlabel("Cluster")
plt.ylabel("Percentage")
plt.title("Income Distribution Across Clusters")
plt.xticks(rotation=0)
plt.legend(title="Income")
plt.show()
```
